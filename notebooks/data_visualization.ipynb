```python
import sys
import os

current_dir = os.getcwd()

project_root = os.path.dirname(current_dir)

if project_root not in sys.path:
    sys.path.append(project_root)
```

```python
from torchvision.datasets import MNIST
import matplotlib.pyplot as plt
DATA_FOLDER = './data'
```

# Data Visualization

```python
train_dataset = MNIST(DATA_FOLDER, train=True, download=True)
```

```python

# Create a figure to hold the plots
fig = plt.figure(figsize=(9, 9))

# Loop 9 times to plot 9 images
for i in range(9):
    # This creates a subplot in a 3x3 grid at position i+1
    plt.subplot(3, 3, i + 1)
    
    # Get the image and label from the dataset
    # train_dataset[i] returns a tuple (image, label)
    image, label = train_dataset[i]
    
    # Plot the image
    # 'cmap='gray'' is important for grayscale images
    plt.imshow(image, cmap='gray')
    
    # Add the label as the title for this subplot
    plt.title(f"Label: {label}")
    
    # Hide the axes (the x and y ticks)
    plt.axis('off')

# Adjust layout to prevent titles from overlapping
plt.tight_layout()

# Display the plot
plt.show()
```

# Noisy Visualization

```python
from src.hito3.noises_transforms import AddGaussianNoise,RandomOcclusion,ContrastChange
from torchvision import transforms
```

## Gaussian Noise

```python
preprocessing = transforms.Compose([
    transforms.ToTensor(),
    AddGaussianNoise(std=0.3)
])
train_dataset = MNIST(DATA_FOLDER, train=True, download=True, transform=preprocessing)
```

```python
# Create a figure to hold the plots
fig = plt.figure(figsize=(9, 9))

# Loop 9 times to plot 9 images
for i in range(9):
    # This creates a subplot in a 3x3 grid at position i+1
    plt.subplot(3, 3, i + 1)
    
    # Get the image and label from the dataset
    # train_dataset[i] returns a tuple (image, label)
    image, label = train_dataset[i]
    image = image.squeeze(0).numpy()
    
    # Plot the image
    # 'cmap='gray'' is important for grayscale images
    plt.imshow(image, cmap='gray')
    
    # Add the label as the title for this subplot
    plt.title(f"Label: {label}")
    
    # Hide the axes (the x and y ticks)
    plt.axis('off')

# Adjust layout to prevent titles from overlapping
plt.tight_layout()

# Display the plot
plt.show()
```

## Contrast

```python
from src.hito3.noises_transforms import AddGaussianNoise,RandomOcclusion,ContrastChange
from torchvision import transforms
import numpy as np
```

```python
preprocessing = transforms.Compose([
    transforms.ToTensor(),
    ContrastChange(factor=5)
])
train_dataset = MNIST(DATA_FOLDER, train=True, download=True, transform=preprocessing)
```

```python
# Create a figure to hold the plots
fig = plt.figure(figsize=(9, 9))

# Loop 9 times to plot 9 images
for i in range(9):
    # This creates a subplot in a 3x3 grid at position i+1
    plt.subplot(3, 3, i + 1)
    
    # Get the image and label from the dataset
    # train_dataset[i] returns a tuple (image, label)
    image, label = train_dataset[i]
    image = image.squeeze(0).numpy()
    
    # Plot the image
    # 'cmap='gray'' is important for grayscale images
    plt.imshow(image, cmap='gray', vmin=0, vmax=1)
    
    # Add the label as the title for this subplot
    plt.title(f"Label: {label}")
    
    # Hide the axes (the x and y ticks)
    plt.axis('off')

# Adjust layout to prevent titles from overlapping
plt.tight_layout()

# Display the plot
plt.show()
```
